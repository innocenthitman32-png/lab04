```python
# --- Part 1: Linear Regression (Salary Dataset) ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load Dataset
df_salary = pd.read_csv("Salary_dataset.csv")
print("Salary Dataset Head:")
display(df_salary.head())

# 2. Train-Test Split
from sklearn.model_selection import train_test_split
X_sal = df_salary[['YearsExperience']]
y_sal = df_salary['Salary']
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_sal, y_sal, test_size=0.2, random_state=42)

# 3. Train Linear Regression Model
from sklearn.linear_model import LinearRegression
sal_model = LinearRegression()
sal_model.fit(X_train_s, y_train_s)

print("\nLinear Regression Coefficients:")
print("Slope (coef_):", sal_model.coef_)
print("Intercept:", sal_model.intercept_)

# 4. Predictions & Evaluation
y_pred_s = sal_model.predict(X_test_s)
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print("\nLinear Regression Metrics:")
print("MAE:", mean_absolute_error(y_test_s, y_pred_s))
print("MSE:", mean_squared_error(y_test_s, y_pred_s))
print("RMSE:", np.sqrt(mean_squared_error(y_test_s, y_pred_s)))
print("R² Score:", r2_score(y_test_s, y_pred_s))


# --- Part 2: Logistic Regression (Diabetes Dataset) ---
df_diabetes = pd.read_csv("diabetes.csv")
print("\nDiabetes Dataset Head:")
display(df_diabetes.head())

# 1. Prepare Features & Target
X_dia = df_diabetes[['Age', 'BMI', 'BloodPressure', 'Insulin', 'Glucose', 'DiabetesPedigreeFunction']]
y_dia = df_diabetes['Outcome']

X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(X_dia, y_dia, test_size=0.2, random_state=42)

# 2. Train Logistic Regression Model
from sklearn.linear_model import LogisticRegression
dia_model = LogisticRegression(max_iter=1000)
dia_model.fit(X_train_d, y_train_d)

print("\nLogistic Regression Intercept & Weights:")
print("Intercept:", dia_model.intercept_)
print("Coefficients:", dia_model.coef_)

# 3. Predictions & Evaluation
y_pred_d = dia_model.predict(X_test_d)
from sklearn.metrics import accuracy_score, confusion_matrix
print("\nLogistic Regression Accuracy:", accuracy_score(y_test_d, y_pred_d))
print("Confusion Matrix:\n", confusion_matrix(y_test_d, y_pred_d))
```
